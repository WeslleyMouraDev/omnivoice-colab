### Endpoints Disponíveis:
1. `/prepare_voice`: Envia áudio de referência (3-15s), transcreve automaticamente e retorna `voice_id`.
2. `/generate_tts`: Sintetiza roteiro usando `voice_id`, divide textos longos e entrega `MP3 (192kbps)` + `SRT` sincronizado.
3. `/clone_and_generate`: Executa o fluxo completo em uma única chamada.
